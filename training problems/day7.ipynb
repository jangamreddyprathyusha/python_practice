7.write a pp to read the list of words as input from the user and return the length of each word for the list

```python
size=int(input('Enter the size of list :'))
WordList=[]
for i in range(size):
    val=input('Enter the Word :')
    WordList.append(val)
print("User entered list is :",WordList)
Len=list(map(lambda w:len(w),WordList))
print("Length of the words in the list :",Len)
```

9.write a pp to read list of elements as input from the user & print the square values of list elements.

10.write a pp to read a list of numbers and print the list of even numbers and odd numbers without using loops & conditional statements

```python
num=list(map(int,input('Enter the number :').split()))
even=list(filter(lambda i:i%2==0,num))
odd=list(filter(lambda i:i%2!=0,num))
print(f"User entered list :{num}")
print(f"Even list :{even}")
print(f"odd list :{odd}")
```
