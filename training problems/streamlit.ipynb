```python
import streamlit as st

# Session state for storing students
if "students" not in st.session_state:
    st.session_state.students = {}

# Grade calculation function
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"

st.title("🎓 Student Result Dashboard")

# Input section
col1, col2 = st.columns(2)

with col1:
    name = st.text_input("Student Name")

with col2:
    marks = st.number_input("Marks", min_value=0, max_value=100)

# Add student button
if st.button("Add Student"):
    if name == "":
        st.warning("⚠️ Student name cannot be empty")
    else:
        st.session_state.students[name] = marks
        if marks >= 40:
            st.success("✅ Student Passed")
        else:
            st.error("❌ Student Failed")

# Table header
if st.session_state.students:
    st.subheader("📊 Student Results")

    h1, h2, h3, h4 = st.columns(4)
    h1.write("*Name*")
    h2.write("*Marks*")
    h3.write("*Grade*")
    h4.write("*Result*")

    for student, mark in st.session_state.students.items():
        grade = calculate_grade(mark)
        result = "Pass" if mark >= 40 else "Fail"

        c1, c2, c3, c4 = st.columns(4)
        c1.write(student)
        c2.write(mark)
        c3.write(grade)
        c4.write(result)
```
