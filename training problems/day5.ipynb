```python
string=input('Enter a string :')
alphabet,digit,spacialchar=0,0,0
for ch in string:
    if ch.isalpha():
        alphabet += 1
    elif ch.isdigit():
        digit += 1
    else:
        spacialchar += 1
print(f"Count of Alphabetic characters is {alphabet}")
print(f"Count of Digits is {digit}")
print(f"Count of Special characters is {spacialchar}")
```

```python
string=input('Enter a string :').lower()
vowels,consonants=0,0
for ch in string:
    if ch.isalpha():
        if ch in 'aeiou':
            vowels += 1
        else:
            consonants += 1
print(f"Count of Vowels is {vowels}")
print(f"Count of Consonants is {consonants}")
```

```python
num=int(input('Enter the integer value :'))
factor=0
for i in range(2,num+1):
    if(num%i==0):
        factor+=1
res="Prime number" if(factor==2) else "Not a prime number"
print(f"{num} is {res}")
```
