6.Write a python program to calculate the simple interest by using the parameterized function

```python
def cal_interest(amount,rateofinterest,time):
    interest=(amount*rateofinterest*time)/100
    print(f"simple interest : {interest}")
cal_interest(12000,4,12)
cal_interest(10000,2,3) 
```

7.Write a python program to read the size of list as input from the user and read the list of element as a input from the user and print the user entered list.

```python
size=int(input('Enter the size of the list :'))
num=[]
for i in range(size):
    val=int(input(f"Enter the element at {i} index :"))
    num.append(val)
print("User entered list :")
print(num)
```

8.Write a pp to read list of size of list as input from the user & read the list element as Alphabetic character & print the user enter list and the sorted version of the user enter list.

```python
n = int(input("Enter the size of the list: "))
lst = []
for i in range(n):
    ch = input(f"Enter alphabetic character {i+1}: ")
    lst.append(ch)
print("User entered list:", lst)
sorted_lst = sorted(lst)
print("Sorted list:", sorted_lst)
```
