```python
#Method Overriding
class Engineer:
    def __init__(self):
        pass
    def work(self):
        print("Engineer works.....!")
class SoftwareEngineer(Engineer):
    def __init__(self):
        super().__init__()
    def work(self):
        print("Software Engineer works...!")
class CivilEngineer(Engineer):
    def __init__(self):
        super().__init__()
    def work(self):
        print("Civil Engineer works...!")
e=Engineer()
se=SoftwareEngineer()
c=CivilEngineer()
e.work()
se.work()
c.work()
```

```python
class BankAccount:
    def _init_(self, name, balance):
        self.name = name
        self.balance = float(balance)
    def deposit(self, amount):
        self.balance += float(amount)
    def withdraw(self, amount):
        # 1% transaction fee using lambda
        fee = (lambda x: x * 0.01)(amount)
        total_withdraw = float(amount) + fee
        if self.balance >= total_withdraw:
            self.balance -= total_withdraw
        else:
            print(" Insufficient balance for withdrawal.")
    def display(self):
        print(f"Account Holder: {self.name}")
        print(f"Final Balance: ₹{self.balance:.2f}")
name = input().strip()
initial_balance = float(input())
withdraw_amount = float(input())
deposit_amount = float(input())
account = BankAccount()
account.withdraw()
account.deposit(deposit_amount)
account.display()
```

```python
class Passenger:
    def init(self, name, age, ticket_price):
        self.name = name
        self.age = int(age)
        self.ticket_price = float(ticket_price)
        self.final_price = self.final_bill()
    def final_bill(self):
        discount_func = lambda price: price * 0.5 if self.age < 12 else (price * 0.3 if self.age > 60 else price)
        discount = discount_func(self.ticket_price)
        final_price = self.ticket_price - discount
        return final_price
    def display(self):
        print(f"Passenger Name : {self.name}")
        print(f"Final Bill : {self.final_price:.2f}")
S1 = Passenger(input(), input(), input())
S1.display()
```
