153 left roted sort array

```python
arr=[1,2,3,4,5]
k=2
n=len(arr) #5
res=[0]*len(arr)  #res=[0,0,0,0,0,0]
for i in range(len(arr)):  #0<5
    res[i]=arr[(i+k+n)%n]   
print(res)
```

right roted sort array

```python
arr=[1,2,3,4,5]
k=2
n=len(arr) #5
res=[0]*len(arr)  #res=[0,0,0,0,0,0]
for i in range(len(arr)):  #0<5
    res[i]=arr[(i-k+n)%n]   
print(res)
```
