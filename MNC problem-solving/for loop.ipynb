```python
for i in range(1,6):
    print(i)
```

7.using dictionary---->keys
when you directly loop through a dictionary,the for loop gives you each key one by one
gives keys,not values

```python
student={"name":"prathu",
         "age":21
}
for key in student:
    print(key)
```

8.both keys & values:---->k+v
.items() allows the for loop to get both the key and its corresponding values

```python
student={"name":"prathu",
         "age":21
}
for key,value in student.items():
    print(key,value)
```

9.using enumerate()--->i+v
give you both the index and value while looping through a sequence.
1.where the item is-->index
2.what the item is-->value


```python
#without enumerate()
names=["A","B","C"]
for name in names:
    print(name)
```

```python
#with enumerate()
names=["A","B","C"]
for index,name in enumerate(names):
    print(index,name)
```

10.nested for loop--->a nfl is a placed inside another for loop

```python
for i in range(3):
    for j in range(2):
        print(i,j)
```

```python
n=20
small=0
for i in range(1,n+1):
    if i<small:
        small=i
print(small)
```

sum of digits

```python
n=12345
sum=0
while n>0:
    dig=n%10
    sum=sum+dig
    n=n//10
print(sum)
```
