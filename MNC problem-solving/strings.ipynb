find max vowels in substring

```python
s="abciiidef"
k=3
max1=0
for i in range(len(s)-2):
    count=0
    for j in range(i,k+i):
        if(s[j]=="a" or s[j]=="e" or s[j]=="i" or s[j]=="o" or s[j]=="u"):
            count+=1
    max1=max(max1,count)
print(max1)
```

how many prime numbers have in factors of 28

```python
n=28
for i in range(1,n+1):
    if n%i==0:
        count=0
        for j in range(1,i+1):
            if i%j==0:
               count+=1
        if count==2:
            print(i,end=" ")
```

```python
class Solution(object):
    def searchRange(self, nums, target):
        def find_first():
            val=0
            left=0
            right=len(nums)-1
            while(left<=right):
                mid=(left+right)//2
                if (nums[mid]==target):
                   val=mid
                   right=mid-1
                elif(nums[mid]<target):
                   left=mid+1
                else:
                   right=mid-1
            return val
        def find_last():
            val=0
            left=0
            right=len(nums)-1
            while(left<=right):
                mid=(left+right)//2
                if (nums[mid]==target):
                   val=mid
                   left=mid+1
                elif(nums[mid]<target):
                   left=mid+1
                else:
                   right=mid-1
            return val
        return[find_first(),find_last()]
```
